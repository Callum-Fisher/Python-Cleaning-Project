```python
import pandas as pd
```

```python
def clean_column_names(df):
    """
    Function Description:
    Standardises column names by converting them to lowercase, removing extra spaces, replacing spaces/hyphens with underscores, and renaming the police data month column to date.

    Input:
    df (DataFrame): Raw dataframe with original column names.

    Output:
    df (DataFrame): Dataframe with standardised column names.
    """
#cleaning column names
    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
        .str.replace("-", "_")
    )
#renaming month column to date
    df = df.rename(
        columns={
            "month": "date"
        }
    )

    return df
```

```python
def clean_dates(df):
    """
    Function Description:
    Converts the date column into datetime format and creates extra date fields for year, month number, and month name.

    Input:
    df (DataFrame): Dataframe containing a date column.

    Output:
    df (DataFrame): Dataframe with year, month_num, and month_name columns added.
    """
#changing date column to date format
    df["date"] = pd.to_datetime(df["date"])
    
#creates a year column
    df["year"] = df["date"].dt.year

#creates month number column
    df["month_num"] = df["date"].dt.month

#creates month name column
    df["month_name"] = df["date"].dt.month_name()

    return df
```

```python
def add_force_name(df, force_name):
    """
    Function Description:
    Adds a police force name to the dataframe so records can be grouped and merged by force.

    Input:
    df (DataFrame): Dataframe to update.
    force_name (str): Name of the police force.

    Output:
    df (DataFrame): Dataframe with police_force column added.
    """

    df["police_force"] = force_name

    return df
```

```python
def select_required_columns(df):
    """
    Function Description:
    Keeps only the columns needed for analysis and removes unnecessary columns to reduce dataset size.

    Input:
    df (DataFrame): Raw or cleaned crime dataframe.

    Output:
    df (DataFrame): Dataframe containing only required columns that exist in the input.
    """

#columns we want
    required_columns = [
        "date",
        "crime_type",
        "longitude",
        "latitude"
    ]

    existing_columns = [
        col for col in required_columns
        if col in df.columns
    ]

    df = df[existing_columns]

    return df
```

```python
def print_size_change(dataset_name, original_df, final_df):
    """
    Function Description:
    Prints a validation summary comparing the number of rows before and after cleaning.

    Input:
    dataset_name (str): Name/path of the dataset being processed.
    original_df (DataFrame): Dataframe before cleaning.
    final_df (DataFrame): Dataframe after cleaning.

    Output:
    None: Prints original row count, final row count, and rows removed.
    """
    original_rows = len(original_df)
    final_rows = len(final_df)
    removed_rows = original_rows - final_rows

    print(f"\n{dataset_name}")
    print(f"Original rows: {original_rows}")
    print(f"Final rows: {final_rows}")
    print(f"Rows removed: {removed_rows}")
```

```python
def process_crime_file(file_path, force_name):
    """
    Function Description:
    Imports and cleans one crime CSV file using the project cleaning functions.

    Input:
    file_path (str): Path to the crime CSV file.
    force_name (str): Name of the police force linked to the file.

    Output:
    df (DataFrame): Cleaned crime dataframe for one file.
    """

#importing csv
    df = pd.read_csv(
        file_path,
        low_memory=False
    )
    
    original_df = df.copy()
    
#cleaning data
    df = clean_column_names(df)

    df= select_required_columns(df)

    df = clean_dates(df)

    #df = filter_crime_types(df)

    df = add_force_name(
        df,
        force_name
    )
    
    print_size_change(
    file_path,
    original_df,
    df
    )

    return df
```

```python
def process_crime_files(crime_files):
    """
    Function Description:
    Processes multiple crime CSV files and combines them into one cleaned dataframe.

    Input:
    crime_files (list): List of tuples containing file paths and police force names.

    Output:
    combined_df (DataFrame): Combined cleaned crime dataframe.
    """

    cleaned_files = []

    for file_path, force_name in crime_files:

        cleaned_df = process_crime_file(
            file_path,
            force_name
        )

        cleaned_files.append(cleaned_df)

#combining all the cleaned crime month dataframes into year dataframes
    combined_df = pd.concat(
        cleaned_files,
        ignore_index=True
    )

    return combined_df
```

```python
def combine_cleaned_years(yearly_dfs):
    """
    Function Description:
    Combines cleaned yearly crime dataframes into one full crime dataset.

    Input:
    yearly_dfs (list): List of cleaned yearly dataframes.

    Output:
    combined_df (DataFrame): Combined dataframe containing all years.
    """
    combined_df = pd.concat(
        yearly_dfs,
        ignore_index=True
    )

    return combined_df
```

```python
def aggregate_crime_monthly(df):
    """
    Function Description:
    Aggregates cleaned crime records to the reporting grain: police force, month, and crime type.

    Input:
    df (DataFrame): Cleaned crime dataframe at individual-record level.

    Output:
    crime_monthly (DataFrame): Monthly aggregated crime dataframe with crime_count.
    """
    crime_monthly = (
        df.groupby(
            [
                "police_force",
                "date",
                "year",
                "month_num",
                "month_name",
                "crime_type"
            ],
            as_index=False
        )
        .size()
        .rename(columns={"size": "crime_count"})
    )

    return crime_monthly
```

```python
def validate_reporting_grain(df):
    """
    Function Description:
    Checks whether the crime reporting dataset has duplicate rows at the intended reporting grain.

    Input:
    df (DataFrame): Monthly aggregated crime dataframe.

    Output:
    None: Prints duplicate count and validation message.
    """
    duplicates = df.duplicated(
        subset=[
            "police_force",
            "date",
            "crime_type"
        ]
    ).sum()

    print("Duplicate reporting rows:", duplicates)

    if duplicates == 0:
        print("Reporting grain is valid.")
    else:
        print("There are duplicate rows.")
```

```python
def check_missing_values(df):
    """
    Function Description:
    Checks and prints the number of missing values in each column.

    Input:
    df (DataFrame): Dataframe to validate.

    Output:
    missing_values (Series): Count of missing values by column.
    """
    missing_values = df.isna().sum()

    print(missing_values)

    return missing_values
```

```python
def export_dataset(df, file_name):
    """
    Function Description:
    Exports a dataframe to CSV without the index.

    Input:
    df (DataFrame): Dataframe to export.
    file_name (str): Output CSV filename or path.

    Output:
    None: Exports CSV file and prints confirmation.
    """
    df.to_csv(
        file_name,
        index=False
    )

    print(f"Exported: {file_name}")
```

```python
def create_monthly_crime_dataset(yearly_dfs, output_file):
    """
    Function Description:
    Runs the monthly crime dataset creation process by combining yearly data, aggregating, validating, checking missing values, and exporting.

    Input:
    yearly_dfs (list): List of cleaned yearly crime dataframes.
    output_file (str): Output CSV filename or path.

    Output:
    crime_monthly (DataFrame): Monthly aggregated crime dataset.
    """

#combines crime data that are in years into one dataframe
    crime_clean = combine_cleaned_years(yearly_dfs)

#aggregate data
    crime_monthly = aggregate_crime_monthly(crime_clean)

#validate data
    validate_reporting_grain(crime_monthly)

#check for missing values
    check_missing_values(crime_monthly)
    
#export the dataframe
    export_dataset(
        crime_monthly,
        output_file
    )

    return crime_monthly
```

```python
#used to correlate the team to the police force
TEAM_FORCE_MAPPING = {
    "Arsenal": "Metropolitan Police",
    "Chelsea": "Metropolitan Police",
    "Tottenham": "Metropolitan Police",
    "West Ham": "Metropolitan Police",
    "Crystal Palace": "Metropolitan Police",

    "Liverpool": "Merseyside Police",
    "Everton": "Merseyside Police",

    "Aston Villa": "West Midlands Police",
    "Wolves": "West Midlands Police",

    "Newcastle": "Northumbria Police"
}
```

```python
def clean_football_columns(df):
    """
    Function Description:
    Standardises football dataset column names.

    Input:
    df (DataFrame): Raw football dataframe.

    Output:
    df (DataFrame): Football dataframe with cleaned column names.
    """
    df = df.copy()

    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
    )

    return df
```

```python
def select_football_columns(df):
    """
    Function Description:
    Keeps only the football columns required for the match-day enrichment analysis.

    Input:
    df (DataFrame): Raw football dataframe.

    Output:
    df (DataFrame): Reduced football dataframe containing required columns.
    """

#columns we need
    required_columns = [
        "date",
        "hometeam",
        "awayteam",
        "fthg",
        "ftag",
        "ftr"
    ]

    existing_columns = [
        col for col in required_columns
        if col in df.columns
    ]

    df = df[existing_columns]

    return df
```

```python
def clean_football_dates(df):
    """
    Function Description:
    Converts football match dates and creates year, month number, and month name fields.

    Input:
    df (DataFrame): Football dataframe containing a date column.

    Output:
    df (DataFrame): Football dataframe with date fields added.
    """
    df = df.copy()

#formats date column to datetime
    df["date"] = pd.to_datetime(
        df["date"],
        dayfirst=True,
        errors="coerce"
    )

#creates year column
    df["year"] = df["date"].dt.year

#creates month number column
    df["month_num"] = df["date"].dt.month

#creates month name column
    df["month_name"] = df["date"].dt.month_name()

    return df
```

```python
def handle_football_missing_values(df):
    """
    Function Description:
    Removes football rows missing essential match identifiers.

    Input:
    df (DataFrame): Football dataframe.

    Output:
    df (DataFrame): Football dataframe with unusable rows removed.
    """

#drops Nans as data becomes useless in the football data if the date, or teams do not exist
    df = df.dropna(
        subset=[
            "date",
            "hometeam",
            "awayteam"
        ]
    )

    return df
```

```python
def filter_relevant_home_teams(df):
    """
    Function Description:
    Keeps only home matches involving teams mapped to the selected police forces.

    Input:
    df (DataFrame): Football dataframe containing hometeam column.

    Output:
    df (DataFrame): Filtered football dataframe for relevant home teams.
    """
    df = df.copy()

    relevant_teams = list(TEAM_FORCE_MAPPING.keys())

    df = df[df["hometeam"].isin(relevant_teams)]

    return df
```

```python
def add_police_force_to_matches(df):
    """
    Function Description:
    Maps each home team to its corresponding police force.

    Input:
    df (DataFrame): Football dataframe containing hometeam column.

    Output:
    df (DataFrame): Football dataframe with police_force column added.
    """
    df = df.copy()

    df["police_force"] = df["hometeam"].map(
        TEAM_FORCE_MAPPING
    )

    return df
```

```python
def create_match_count_column(df):
    """
    Function Description:
    Creates a match-count indicator so home matches can be aggregated by month.

    Input:
    df (DataFrame): Cleaned football dataframe.

    Output:
    df (DataFrame): Football dataframe with home_match_count column added.
    """
    df = df.copy()

    df["home_match_count"] = 1

    return df
```

```python
def create_derby_column(df):
    """
    Function Description:
    Flags selected derby matches for event-based analysis.

    Input:
    df (DataFrame): Football dataframe containing hometeam and awayteam columns.

    Output:
    df (DataFrame): Football dataframe with derby_match_count column added.
    """
    df = df.copy()

#derby matches in the data we have
    derby_matches = [
        ("Liverpool", "Everton"),
        ("Arsenal", "Tottenham")
    ]

    df["derby_match_count"] = 0

    for team_a, team_b in derby_matches:
        derby_mask = (
            ((df["hometeam"] == team_a) & (df["awayteam"] == team_b)) |
            ((df["hometeam"] == team_b) & (df["awayteam"] == team_a))
        )

        df.loc[derby_mask, "derby_match_count"] = 1

    return df
```

```python
def validate_reporting_grain_football(df):
    """
    Function Description:
    Checks whether the aggregated football dataset has duplicate rows at police force, year, and month level.

    Input:
    df (DataFrame): Aggregated football dataframe.

    Output:
    None: Prints duplicate count and validation message.
    """
    duplicates = df.duplicated(
        subset=[
            "police_force",
            "year",
            "month_num"
        ]
    ).sum()

    print("Duplicate reporting rows:", duplicates)

    if duplicates == 0:
        print("Reporting grain is valid.")
    else:
        print("There are duplicate rows.")
```

```python
def process_football_file(file_path):
    """
    Function Description:
    Imports and cleans one football CSV file using the football cleaning pipeline.

    Input:
    file_path (str): Path to the football CSV file.

    Output:
    df (DataFrame): Cleaned football dataframe for one file.
    """
#imports files
    df = pd.read_csv(
        file_path,
        low_memory=False
    )
    original_df = df.copy()
    
    df = clean_football_columns(df)
    df = select_football_columns(df)
    df = clean_football_dates(df)
    df = handle_football_missing_values(df)
    df = filter_relevant_home_teams(df)
    df = add_police_force_to_matches(df)
    df = create_match_count_column(df)
    df = create_derby_column(df)

    print_size_change(
    file_path,
    original_df,
    df
    )

    return df
```

```python
def process_football_files(football_files):
    """
    Function Description:
    Processes multiple football CSV files and combines them into one cleaned dataframe.

    Input:
    football_files (list): List of football CSV file paths.

    Output:
    football_clean (DataFrame): Combined cleaned football dataframe.
    """
    cleaned_files = []

    for file_path in football_files:
        cleaned_df = process_football_file(file_path)
        cleaned_files.append(cleaned_df)

    football_clean = pd.concat(
        cleaned_files,
        ignore_index=True
    )

    return football_clean
```

```python
def aggregate_football_monthly(
    df,
    output_file
):
    """
    Function Description:
    Aggregates cleaned football data to police force and month level, validates it, checks missing values, and exports it.

    Input:
    df (DataFrame): Cleaned football dataframe.
    output_file (str): Output CSV filename or path.

    Output:
    football_monthly (DataFrame): Aggregated football dataframe.
    """

    football_monthly = (
        df.groupby(
            [
                "police_force",
                "year",
                "month_num",
                "month_name"
            ],
            as_index=False
        )
        .agg({
            "home_match_count": "sum",
            "derby_match_count": "sum"
        })
    )
#validates data
    validate_reporting_grain_football(football_monthly)
    
#checks for missing values
    check_missing_values(football_monthly)

#creates a csv file for the cleaned football data
    football_monthly.to_csv(
        output_file,
        index=False
    )

    print(f"Exported: {output_file}")

    return football_monthly
```

```python
def merge_crime_and_football(
    crime_df,
    football_df
):
    """
    Function Description:
    Merges monthly crime data with monthly football activity data.

    Input:
    crime_df (DataFrame): Monthly aggregated crime dataframe.
    football_df (DataFrame): Monthly aggregated football dataframe.

    Output:
    final_df (DataFrame): Crime dataframe enriched with football match metrics.
    """

#left merge on football data
    final_df = crime_df.merge(
        football_df,
        on=[
            "police_force",
            "year",
            "month_num",
            "month_name"
        ],
        how="left"
    )

# Replace missing football values with 0
    final_df["home_match_count"] = (
        final_df["home_match_count"]
        .fillna(0)
        .astype(int)
    )

    final_df["derby_match_count"] = (
        final_df["derby_match_count"]
        .fillna(0)
        .astype(int)
    )

#validating dataframe
    duplicates = final_df.duplicated(
        subset=[
            "police_force",
            "year",
            "month_num",
            "crime_type"
        ]
    ).sum()

    print(
        f"Duplicate rows: {duplicates}"
    )

#checks for missing values
    check_missing_values(final_df)

    return final_df
```

```python
def load_population_data(file_path):
    """
    Function Description:
    Imports the population CSV file using the correct header row.

    Input:
    file_path (str): Path to the population CSV file.

    Output:
    df (DataFrame): Raw population dataframe.
    """
#imports csv
    df = pd.read_csv(
        file_path,
        header=1
    )

    return df
```

```python
def clean_population_columns(df):
    """
    Function Description:
    Standardises population dataset column names.

    Input:
    df (DataFrame): Raw population dataframe.

    Output:
    df (DataFrame): Population dataframe with cleaned column names.
    """
    df = df.copy()

    df.columns = (
        df.columns
        .str.lower()
        .str.strip()
        .str.replace(" ", "_")
    )

    return df
```

```python
def select_population_columns(df):
    """
    Function Description:
    Keeps only the local authority name and yearly population columns required for the project.

    Input:
    df (DataFrame): Raw population dataframe.

    Output:
    df (DataFrame): Reduced population dataframe.
    """
#columns we need in the population data
    required_columns = [
        "laname23",
        "population_2019",
        "population_2020",
        "population_2021",
        "population_2022",
        "population_2023",
        "population_2024"
    ]

    df = df[required_columns]

    return df
```

```python
#used to correlate the area with the police force
POLICE_FORCE_AREAS = {
    "Metropolitan Police": [
        "Barking and Dagenham", "Barnet", "Bexley", "Brent", "Bromley",
        "Camden", "Croydon", "Ealing", "Enfield", "Greenwich",
        "Hackney", "Hammersmith and Fulham", "Haringey", "Harrow",
        "Havering", "Hillingdon", "Hounslow", "Islington",
        "Kensington and Chelsea", "Kingston upon Thames", "Lambeth",
        "Lewisham", "Merton", "Newham", "Redbridge",
        "Richmond upon Thames", "Southwark", "Sutton",
        "Tower Hamlets", "Waltham Forest", "Wandsworth",
        "Westminster", "City of London"
    ],

    "Merseyside Police": [
        "Liverpool", "Knowsley", "Sefton", "St. Helens", "Wirral"
    ],

    "West Midlands Police": [
        "Birmingham", "Coventry", "Dudley", "Sandwell",
        "Solihull", "Walsall", "Wolverhampton"
    ],

    "Northumbria Police": [
        "Newcastle upon Tyne", "Gateshead", "North Tyneside",
        "South Tyneside", "Sunderland", "Northumberland"
    ]
}
```

```python
def add_police_force_from_local_authority(df):
    """
    Function Description:
    Maps local authorities to the police forces used in the project and removes unmapped areas.

    Input:
    df (DataFrame): Population dataframe at local-authority level.

    Output:
    df (DataFrame): Population dataframe with police_force column added.
    """
    df = df.copy()

    area_to_force = {}

    for force, areas in POLICE_FORCE_AREAS.items():
        for area in areas:
            area_to_force[area] = force

    df["police_force"] = df["laname23"].map(area_to_force)

    df = df.dropna(subset=["police_force"])

    return df
```

```python
def reshape_population_data(df):
    """
    Function Description:
    Converts population data from wide format to long format so each row represents one police force and year.

    Input:
    df (DataFrame): Population dataframe with one population column per year.

    Output:
    population_long (DataFrame): Long-format population dataframe with year and population columns.
    """
    population_long = df.melt(
        id_vars="police_force",
        value_vars=[
            "population_2019",
            "population_2020",
            "population_2021",
            "population_2022",
            "population_2023",
            "population_2024"
        ],
        var_name="year",
        value_name="population"
    )

    return population_long
```

```python
def clean_population_year(df):
    """
    Function Description:
    Converts population year labels into integer year values.

    Input:
    df (DataFrame): Long-format population dataframe with year labels.

    Output:
    df (DataFrame): Population dataframe with numeric year column.
    """
    df = df.copy()

    df["year"] = (
        df["year"]
        .str.replace("population_", "")
        .astype(int)
    )

    return df
```

```python
def clean_population_values(df):
    """
    Function Description:
    Converts population values to integers after removing commas.

    Input:
    df (DataFrame): Population dataframe containing population values.

    Output:
    df (DataFrame): Population dataframe with numeric population column.
    """
    df = df.copy()

    df["population"] = (
        df["population"]
        .astype(str)
        .str.replace(",", "")
        .astype(int)
    )

    return df
```

```python
def validate_population_data(df):
    """
    Function Description:
    Validates population data by checking duplicates, missing values, and police force counts.

    Input:
    df (DataFrame): Population dataframe to validate.

    Output:
    None: Prints validation summary.
    """
    duplicates = df.duplicated(
        subset=[
            "police_force",
            "year"
        ]
    ).sum()

    print("Duplicate rows:", duplicates)

    print("\nMissing values:")
    print(df.isna().sum())

    print("\nPolice forces:")
    print(df["police_force"].value_counts())
```

```python
def export_population_data(df, output_file):
    """
    Function Description:
    Exports the cleaned population dataframe to CSV.

    Input:
    df (DataFrame): Population dataframe to export.
    output_file (str): Output CSV filename or path.

    Output:
    None: Exports CSV file and prints confirmation.
    """
#exporting csv file of population dataframe
    df.to_csv(
        output_file,
        index=False
    )

    print(f"Exported: {output_file}")
```

```python
def aggregate_population_by_force(df):
    """
    Function Description:
    Aggregates local-authority population values to police-force and year level.

    Input:
    df (DataFrame): Population dataframe at local-authority/year level.

    Output:
    population_by_force (DataFrame): Population dataframe aggregated to police-force/year level.
    """

#aggregating population data
    population_by_force = (
        df.groupby(
            ["police_force", "year"],
            as_index=False
        )
        .agg(
            population=("population", "sum")
        )
    )

    return population_by_force
```

```python
def add_2025_population(df):
    """
    Function Description:
    Creates 2025 population rows by reusing 2024 values as proxy estimates.

    Input:
    df (DataFrame): Population dataframe containing 2024 rows.

    Output:
    df (DataFrame): Population dataframe with 2025 rows added.
    """

#creating 2025 population column
    population_2025 = (
        df[df["year"] == 2024]
        .copy()
    )

    population_2025["year"] = 2025

#adding the column to the datafarme
    df = pd.concat(
        [df, population_2025],
        ignore_index=True
    )

    return df
```

```python
def process_population_data(file_path, output_file):
    """
    Function Description:
    Runs the full population pipeline: import, clean, map areas, reshape, clean values, validate, aggregate, add 2025, and export.

    Input:
    file_path (str): Path to population CSV file.
    output_file (str): Output CSV filename or path.

    Output:
    df (DataFrame): Cleaned and aggregated population dataframe.
    """
    df = load_population_data(file_path)

    df = clean_population_columns(df)

    df = select_population_columns(df)

    df = add_police_force_from_local_authority(df)

    df = reshape_population_data(df)

    df = clean_population_year(df)

    df = clean_population_values(df)

    validate_population_data(df)

    df = aggregate_population_by_force(df)

    df = add_2025_population(df)

    export_population_data(
        df,
        output_file
    )

    return df
```
